## Problem and input/output contract

Platforms coordinate eligibility, ranking, transactions, and feedback. Contracts pre-agree inventory and delivery obligations; auctions compete per request. Search describes the demand entry point, native describes presentation, and programmatic describes trading. These dimensions overlap. Advertising is more than CTR prediction.
