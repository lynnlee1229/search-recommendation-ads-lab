## Problem and input/output contract

Define CTR=clicks/impressions, CVR=conversions/clicks, and CTCVR=conversions/impressions. CTCVR=CTR×CVR requires conversions to belong to the click path. CPM=1000×spend/impressions, CPC=spend/clicks, CPA=spend/conversions. eCPM is expected currency per thousand impressions; for a CPC bid, eCPM=1000×pCTR×CPC. ROAS=attributed revenue/ad spend, not profit margin. This example has no platform fee, so publisher receipts equal advertiser payments.

Before running, predict which result the changed variable will affect and which invariants must hold.

## Run the one-variable comparison

Inputs are small constructed examples visible in source. Output units and constraints are defined above. Run assertions to verify mathematics and accounting.

```python
a=ac.billing(); b=ac.billing(clicks=400)
display(pd.DataFrame([a,b],index=['200 clicks','400 clicks']))
assert abs(a['ctr']*a['cvr']-a['ctcvr'])<1e-12
assert abs(1000*a['ctr']*a['cpc']-a['cpm'])<1e-12
assert a['publisher_receipt']==80
```

## Interpretation and limits

Changing only click count while spend and conversions stay fixed increases CTR, reduces CPC and CVR, and leaves CPA and ROAS unchanged. Higher CTR does not establish higher advertiser profit. Revenue also differs from incremental revenue caused by advertising.
